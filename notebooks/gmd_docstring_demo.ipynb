## 2. Sanity checks (offline)

```python
doc = gmd.__doc__

assert doc is not None and doc.strip(), "docstring missing"

for section in ("Args:", "Returns:", "Raises:", "Examples:"):
    assert section in doc, f"missing section: {section}"

for arg in ("variables", "country", "version", "raw", "vars",
            "sources", "cite", "start_year", "end_year"):
    assert arg in doc, f"argument not documented: {arg}"

print("OK — docstring present, all sections and key arguments documented.")
```

## 3. Try the documented examples (needs internet)

These mirror the `Examples:` block in the docstring. They download data, so they only work online.

```python
# Print every available data vintage (prints, returns None).
gmd(version="list")
```

```python
# Print the variable definitions.
gmd(vars="list")
```

```python
# Filter to specific countries and variables for one vintage.
df = gmd(
    version="2025_12",
    country=["USA", "CHN"],
    variables=["rGDP", "infl", "unemp"],
)
df.head()
```

```python
# Restrict the year range and cache the download locally.
df = gmd(variables="rGDP", start_year=1990, end_year=2020, fast=True)
df.head()
```

```python
# Print the BibTeX citation for the database.
gmd(cite="GMD")
```
